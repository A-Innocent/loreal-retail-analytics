```python
import pandas as pd

folder = r"C:\Users\ainno\PycharmProjects\JupyterProject\data"
master = pd.read_csv(f"{folder}/sephora_master.csv", parse_dates=["submission_time"], low_memory=False)
print("Loaded:", master.shape)
```

```python
output = r"C:\Users\ainno\PycharmProjects\JupyterProject\data"

agg_brand = master.groupby(["brand_name", "primary_category"]).agg(
    avg_review_rating=("review_rating", "mean"),
    avg_product_rating=("product_rating", "mean"),
    total_reviews=("review_rating", "count"),
    avg_price=("price_usd", "mean"),
    total_loves=("loves_count", "max"),
    pct_recommended=("is_recommended", "mean")
).reset_index().round(2)

agg_brand.to_csv(f"{output}/agg_brand.csv", index=False)
print("agg_brand:", agg_brand.shape)

agg_category = master.groupby(
    ["primary_category", "secondary_category", "review_year", "review_month"]
).agg(
    avg_rating=("review_rating", "mean"),
    total_reviews=("review_rating", "count"),
    avg_price=("price_usd", "mean"),
    pct_recommended=("is_recommended", "mean")
).reset_index().round(2)

agg_category.to_csv(f"{output}/agg_category.csv", index=False)
print("agg_category:", agg_category.shape)

agg_skin = master[master["skin_type"].notna()].groupby(
    ["skin_type", "primary_category", "secondary_category"]
).agg(
    avg_rating=("review_rating", "mean"),
    total_reviews=("review_rating", "count"),
    pct_recommended=("is_recommended", "mean")
).reset_index().round(2)

agg_skin.to_csv(f"{output}/agg_skin.csv", index=False)
print("agg_skin:", agg_skin.shape)

agg_time = master.groupby(["review_year", "review_month", "primary_category"]).agg(
    avg_rating=("review_rating", "mean"),
    total_reviews=("review_rating", "count"),
    pct_recommended=("is_recommended", "mean")
).reset_index().round(2)

agg_time.to_csv(f"{output}/agg_time.csv", index=False)
print("agg_time:", agg_time.shape)

print("\nAll aggregated tables exported.")
```
